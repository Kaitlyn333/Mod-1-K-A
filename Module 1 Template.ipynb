Prevalence and Incidence

Alzheimer's is a cognitive disease that causes loss of memory and cognitive functions and typically worsens over time. Between 2000 and 2010, deaths attributed to Alzheimer's rose 68%, and more recent Alzheimer's Association reporting notes reported deaths from Alzheimer's disease rose 134% between 2000 and 2024. New-case incidence rises steeply with age too. Historical Alzheimer's Association projections estimated more than 238,000 new cases among people over 85 in a single year, more than the 65–74 and 75–84 age groups combined. The incidence of Alzheimer’s disease continues to rise as boomers age.


World: More than 55 million people worldwide live with dementia, of which Alzheimer's disease accounts for an estimated 60% to 70% of cases, putting global Alzheimer's cases in the range of roughly 33–38 million. The number of dementia patients is projected to rise to 78 million by 2030 and 139 million by 2050 as populations age. More than 60% of Alzheimer's patients live in low and middle income countries, and about 10 million new cases are added each year globally.


United States: Alzheimer's is a cognitive disease that causes loss of memory and cognitive functions and typically worsens over time. Alzheimer's is more likely for older people, specifically above the age of 65 although it can happen earlier, denoting it as early onset Alzheimer's. It is estimated that 7.4 Million Americans above the age of 65 are currently living with Alzheimer’s and 74% are age 75 and older. Nearly two thirds of Alzheimer's patients are female.

Economic Burden

Alzheimer's disease imposes a substantial financial burden on patients, families, and health systems, with costs coming from medical care, long-term care, and unpaid caregiving.

World: Global cost estimates from Alzheimer's Disease International have placed the annual worldwide cost of dementia at roughly $818 billion as of 2015, up 35% from about $604 billion in 2010, with projections that it could approach $2 trillion by 2050 as the number of people affected continues to grow. More than half of these costs occur in high-income countries, even though most people with dementia live in low- and middle-income countries.

United States: In 2026, health and long-term care costs for people living with Alzheimer's and other dementias are projected to reach $409 billion, not including the value of unpaid caregiving; Medicare and Medicaid are expected to cover $263 billion (64%), while out-of-pocket spending is expected to be $103 billion. A broader federally funded estimate from the USC Schaeffer Center, which also accounts for lost quality of life and unpaid family caregiving, puts the total 2026 cost of dementia at $818 billion, with $237 billion of that coming from unpaid caregivers who provide 6.8 billion hours of care annually. Under this broader measure, families and individuals bear roughly 80% of the total cost, more than three times the burden carried by health systems and government programs combined.

Risk Factors
Genetic: Age is the strongest overall risk factor, and having a parent or sibling with Alzheimer's raises risk. The best-established genetic risk factor is the APOE ε4 allele. Carrying one copy increases risk roughly two- to three-fold, and two copies raises it considerably more, though it is not deterministic, and many carriers never develop the disease. Rare, inherited mutations in the APP, PSEN1, and PSEN2 genes cause early-onset familial Alzheimer's, which accounts for a very small share of cases.
Lifestyle and modifiable: The 2024 Lancet Commission identified 14 modifiable risk factors, and estimated 45% of future dementia could potentially be prevented if all were eliminated. These include low education, hearing loss, high LDL cholesterol, depression, traumatic brain injury, physical inactivity, diabetes, smoking, hypertension, obesity, excessive alcohol consumption, social isolation, air pollution, and untreated vision loss. The relevance of each varies across the lifespan: education matters in early life, several vascular and metabolic factors in midlife, and social isolation, air pollution, and vision loss in late life. Cardiovascular health in particular is closely linked to brain health, so what is good for the heart is generally good for the brain.
Gene-environment interaction: Genetics and lifestyle do not act independently. Research suggests the APOE ε4 allele may modify the effects of environmental exposures, so that different modifiable factors, such as diabetes or hearing loss, may matter more depending on carrier status. Overall, non-modifiable risks like age and genetics cannot be changed, but lifestyle factors offer real opportunities for prevention or delay.
