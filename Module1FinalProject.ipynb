**Risk Factors:**
Genetic: Age is the strongest overall risk factor, and having a parent or sibling with Alzheimer's raises risk. The best-established genetic risk factor is the APOE ε4 allele. Carrying one copy increases risk roughly two- to three-fold, and two copies raises it considerably more, though it is not deterministic, and many carriers never develop the disease. Rare, inherited mutations in the APP, PSEN1, and PSEN2 genes cause early-onset familial Alzheimer's, which accounts for a very small share of cases.
Lifestyle and modifiable: The 2024 Lancet Commission identified 14 modifiable risk factors, and estimated 45% of future dementia could potentially be prevented if all were eliminated. These include low education, hearing loss, high LDL cholesterol, depression, traumatic brain injury, physical inactivity, diabetes, smoking, hypertension, obesity, excessive alcohol consumption, social isolation, air pollution, and untreated vision loss. The relevance of each varies across the lifespan: education matters in early life, several vascular and metabolic factors in midlife, and social isolation, air pollution, and vision loss in late life. Cardiovascular health in particular is closely linked to brain health, so what is good for the heart is generally good for the brain.
Gene-environment interaction: Genetics and lifestyle do not act independently. Research suggests the APOE ε4 allele may modify the effects of environmental exposures, so that different modifiable factors, such as diabetes or hearing loss, may matter more depending on carrier status. Overall, non-modifiable risks like age and genetics cannot be changed, but lifestyle factors offer real opportunities for prevention or delay.

**Disease progression & prognosis:**
Alzheimer’s is a progressive neurodegenerative disease, which essentially means that symptoms gradually worsen over time. In the early stage, patients experience noticeable memory lapses, such as losing objects or struggling to find the right words. As it advances to the middle stage, patients experience deeper confusion, significant personality changes, and start to require help with daily activities. During the late stage, there is a loss of ability to communicate, swallow, and control basic motor functions. Ultimately, the disease is fatal, with an average life expectancy of 4 to 8 years following a formal diagnosis.

**Biological mechanisms (anatomy, organ physiology, cell & molecular physiology):**
The entorhinal cortex is responsible for holding past memories, and it is the first part of the brain affected by Alzheimer's, followed by the hippocampus, which is responsible for creating new memories. At the molecular level, amyloid beta is typically located at the synapses of a healthy brain; however, Alzheimer's leads to the development of amyloid beta with varying amino acid sequence lengths. Some studies suggest that the disease causes an increased creation of amyloid beta 42 rather than amyloid beta 40, which might serve as an early symptom. During Alzheimer’s, an amyloid beta “plaque” begins to develop in between synapses. Unlike amyloid beta, tau is located inside the cell itself. During Alzheimer’s, tau—similar to amyloid beta—starts to become “sticky” and forms tau “tangles” around the axon, which impedes the flow of neurotransmitters down the neuron’s axon. Additionally, an increase of pTau (phosphorylated tau) has been observed during the disease. Overall, the brain shrinks due to different factors such as lack of blood flow and insufficient glucose levels needed to keep it functioning properly.

## Data Analysis: 


```python
#These libraries below were used in the analyis of our data, along with the "Patient" class from the patients file, which can be accessed in the GitHub linked below
from patients import Patient
import matplotlib.pyplot as plt     
from scipy import stats             
import numpy as np                  
import statistics                   
```

The below code instantiates our data using our Patient class, which makes use of a Pandas dataframe. Then, the values for brain weights for males and females, respectively, are put into two different lists, which will be manipulated in order to produce the mean brain weights for both groups.

```python
# Load all patient records from the CSV file into Patient.all_patients
Patient.instantiate_from_csv("Metadata and Protein Data for Module 1.csv")

# Holds brain weights of male patients
male_bw = []     
# Holds brain weights of female patients
female_bw = []   


# Loop over every Male patient and collect their brain weight
for Patient in Patient.filter(Patient.all_patients, sex="Male"):
    male_bw.append(Patient.brain_weight)


# Loop over every Female patient and collect their brain weight
for Patient in Patient.filter(Patient.all_patients, sex="Female"):
    female_bw.append(Patient.brain_weight)
```

The below code starts by removing the one and only instance of Unavailable data within the dataset for fresh brain weight before continuing to calculate the mean and standard deviation for fresh brain weight in both groups using scipy

```python
# Drop any "Unavailable" entries so they don't break the math below (not necessary for male_bw)
female_bw = [x for x in female_bw if x != "Unavailable"]


# Compute the mean brain weight for each sex
x_male_bar = (statistics.mean(male_bw))
x_female_bar = (statistics.mean(female_bw))


# Compute the standard deviation of brain weight for each sex
bw_male_stdev = (statistics.stdev(male_bw))
bw_female_stdev = (statistics.stdev(female_bw))
```

```python
# Print the computed stats
print(f'x_male_bar = {x_male_bar}, age_dementia_stdev {bw_male_stdev}')
print(f'x_female_bar = {x_female_bar}, bw_femalestdev {bw_female_stdev}')
```

The below code instantiates and assigns values to variables that will serve to comprise the primary components of our bar graph (including the x axis titles and the error bars on the bar graph)

```python
# Data for the bar chart: one bar per sex with error bars
sex_cols = ['Male', 'Female']
mean_sex = [x_male_bar, x_female_bar]
stdev_sex = [bw_male_stdev, bw_female_stdev]
yerr = [stdev_sex, stdev_sex]
```

The below code runs the two sample t-test on the list of brain weight data of the two groups using the stats library (from scipy) and returns a t value and a p value, these values can be printed out and seen

```python
t_stat, p_val = stats.ttest_ind(male_bw, female_bw)
print(f't_stat = {t_stat}, p_val = {p_val}')
```

The below code uses the matplotlib Python library in order to generate the bar graph for our data analysis. The graph includes the two columns representing our independent variables along with the p value and t value that were calculated for our dataset.

```python
plt.bar(sex_cols, mean_sex, yerr=yerr, capsize=10, color=["blue", "orange"])
plt.title("How Biological Sex Affects Brain Weight")
plt.xlabel("Biological Sex")
plt.ylabel("Brain Weight")
y_max = max(mean_sex) + max(stdev_sex) * .4
plt.text(0.5, y_max - 30, f"t = {t_stat:.2f}\np = {p_val:.3e}", ha='center', va='bottom')
plt.show()
```


Now for the scatterplot. The following code loads the patient data and creates the ABeta42 and pTAU variables.

```python
# Holds ABeta42 levels for every patient
ABeta42 = []   
# Hold pTAU levels for every patient
pTAU = []
```

Below, the code puts all the ABeta42 concentration values for each patient in the list called ABeta42.
And it puts all the pTAU concentration values for each patient in the list called pTAU.

```python
# Collect ABeta42 measurement from each patient
for Patient in Patient.all_patients:
    ABeta42.append(Patient.ABeta42)


# Collect pTAU measurement for every patient
for Patient in Patient.all_patients:
    pTAU.append(Patient.pTAU)
```

Create the independant and dependant variables for the scatterplot.

```python
# Independent variable
X = ABeta42
# Dependent variable
Y = pTAU
```

The following lines calculate the linear regression equation and use it to determine the predicted Y values

```python
# Calculate the regression line
slope, intercept, r_value, p_value, std_err = stats.linregress(X, Y)


# Calculate the predicted Y values
regression_line = slope * np.array(X) + intercept
```

Creates scatterplot, plots the regression line and adds plot and axis titles. Uses the matplotlib Python library in order to generate the scatterplot for our data analysis. Each individual dot accounts for a single patient with a value for ABeta42 and pTAU along the X and Y axis respectively.

```python
# Create scatter plot
plt.scatter(X, Y, color='blue')


# Add regression line
plt.plot(X, regression_line, color='red')


# Add labels and title
plt.xlabel('Amyloid-Beta 42 (pg/ug)')
plt.ylabel('pTAU (pg/ug)')
plt.title('Scatter Plot of ABeta42 levels vs pTAU levels (pg/ug)')
```

Displays the line of regression equation and the R value in the top right hand corner. Adjusts the font and position to the desired place. And finally show the scatterplot!

```python
# Add equation and R value to top-right corner
equation = f'y = {slope:.4f}x + {intercept:.2f}'
r_text = f'R = {r_value:.2f}'

plt.text(
    0.95, 0.95,
    equation + '\n' + r_text,
    transform=plt.gca().transAxes,
    ha='right',
    va='top'
)


# Show scatter plot
plt.show()
```

The below code is an additional bar graph and two-samples t-test representing the relationship between cognitive status (those diagnosed with dementia and those not diagnosed) and amyloid-beta 42 levels in the brain (measured in pg/ug)

```python
# Holds amyloid beta levels of patients with dementia
dementia_AB = []     
# Holds amyloid beta levels of patients without dementia
nondementia_AB = []   


# Loop over every patient with dementia and append their amyloid-beta levels to the list
for Patient in Patient.filter(Patient.all_patients, status="Dementia"):
    dementia_AB.append(Patient.ABeta42)


# Loop over every patient not diagnosed with dementia and append their amyloid-beta levels to the list
for Patient in Patient.filter(Patient.all_patients, status="No dementia"):
    nondementia_AB.append(Patient.ABeta42)

# Compute the mean amyloid-beta levels for each patient with dementia
x_dementia_bar = (statistics.mean(dementia_AB))
x_nondementia_bar = (statistics.mean(nondementia_AB))


# Compute the standard deviation of brain weight for each sex
AB_dementia_stdev = (statistics.stdev(dementia_AB))
AB_nondementia_stdev = (statistics.stdev(nondementia_AB))

# Print the computed stats
print(f'x_dementia_bar = {x_dementia_bar}, AB_dementia_stdev {AB_dementia_stdev}')
print(f'x_nondementia_bar = {x_nondementia_bar}, AB_nondementia_stdev {AB_nondementia_stdev}')

# Data for the bar chart: one bar per IV with error bars
status_cols = ['Dementia', 'No Dementia']
mean_status = [x_dementia_bar, x_nondementia_bar]
stdev_status = [AB_dementia_stdev, AB_nondementia_stdev]
yerr = [(0, 0), stdev_status]

#T values and p values for bar graph
t_stat, p_val = stats.ttest_ind(dementia_AB, nondementia_AB)
print(f't_stat = {t_stat}, p_val = {p_val}')

#Outputs graph, complete with IV, DV, error bars, and statistical analysis values
plt.bar(status_cols, mean_status, yerr=yerr, capsize=10, color=["blue", "orange"])
plt.title("How Cognitive Status Correlates with Amyloid-Beta 42 Levels in the Brain")
plt.xlabel("Cognitive Status")
plt.ylabel("Amyloid-Beta 42 (pg/ug)")
y_max = max(mean_status) + max(stdev_status) * .4
plt.text(0.5, y_max - 30, f"t = {t_stat:.2f}\np = {p_val:.3e}", ha='center', va='bottom')
plt.show()
```

This graph and the associated statistical analysis proves that there is a statistically significant difference concerning Amyloid-Beta 42 levels in patients with dementia compared to patients not diagnosed with dementia. It is important to note that the results of this graph will not be discussed in the Discussion questions below, as we decided to focus on our bar graph concerning the correlation between sex and fresh brain weight.

The below code is an additional scatterplot with a line of best fit and r^2 value representing the relationship between fresh brain weight and pTAU levels

```python
# Holds brain weight for every patient
brain_weight = []   

# Holds pTAU levels for every patient
pTAU = []      


# Filter and collect both measurements in a single loop
for patient in Patient.all_patients:
    # Check that brain_weight is present and valid
    if patient.brain_weight != "Unavailable" and patient.brain_weight is not None:
        brain_weight.append(float(patient.brain_weight))
        pTAU.append(float(patient.pTAU))

# Independent variable
X = brain_weight

# Dependent variable
Y = pTAU



# Calculate the regression line
slope, intercept, r_value, p_value, std_err = stats.linregress(X, Y)
r_squared_value = r_value ** 2


# Calculate the predicted Y values
regression_line = slope * np.array(X) + intercept


# Create scatter plot
plt.scatter(X, Y, color='blue')


# Add regression line
plt.plot(X, regression_line, color='red')


# Add labels and title
plt.xlabel('Fresh Brain Weight (g)')
plt.ylabel('pTAU (pg/ug)')
plt.title('Scatter Plot of Brain Weight (g) vs pTAU levels (pg/ug)')


# Add equation, R^2, and p-value to top-right corner
equation = f'y = {slope:.2f}x + {intercept:.2f}'
r_text = f'R^2 = {r_squared_value:.2f}'

# Format p-value using scientific notation 
p_text = f'p = {p_value:.3e}' 

plt.text(
    0.95, 0.85,
    equation + '\n' + r_text + '\n' + p_text,
    transform=plt.gca().transAxes,
    ha='right',
    va='top'
)


# Show scatter plot
plt.show()
```
